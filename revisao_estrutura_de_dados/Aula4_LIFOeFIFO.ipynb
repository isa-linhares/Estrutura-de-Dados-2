**Exercício 1 - O simulador "desfazer"**

```python
# EXERCÍCIO 1 - PILHA

pilha = []


# Adicionar uma ação na pilha
def adicionar_acao(acao):
    pilha.append(acao)
    print("\nAção registrada com sucesso!")


# Desfazer a última ação
def desfazer():
    if len(pilha) > 0:
        acao = pilha.pop()
        print("\nAção desfeita:", acao)
    else:
        print("\nNão há ações para desfazer.")


# Mostrar o histórico
def mostrar_historico():
    if len(pilha) == 0:
        print("\nO histórico está vazio.")
    else:
        print("\n--- HISTÓRICO DE AÇÕES ---")

        for i, acao in enumerate(pilha, 1):
            print(i, "-", acao)


# Programa principal
while True:

    print("\n==============================")
    print("     SISTEMA HOSPITALAR")
    print("==============================")
    print("1 - Digitar")
    print("2 - Apagar")
    print("3 - Substituir")
    print("4 - Desfazer última ação")
    print("5 - Ver histórico")
    print("6 - Sair")
    print("==============================")

    opcao = input("Escolha uma opção: ")

    # DIGITAR
    if opcao == "1":

        texto = input("Digite a informação do paciente: ")

        acao = "Digitar: " + texto

        adicionar_acao(acao)

    # APAGAR
    elif opcao == "2":

        texto = input("Digite a informação que deseja apagar: ")

        acao = "Apagar: " + texto

        adicionar_acao(acao)

    # SUBSTITUIR
    elif opcao == "3":

        antigo = input("Digite a informação que deseja substituir: ")
        novo = input("Digite a nova informação: ")

        acao = "Substituir '" + antigo + "' por '" + novo + "'"

        adicionar_acao(acao)

    # DESFAZER
    elif opcao == "4":

        desfazer()

    # HISTÓRICO
    elif opcao == "5":

        mostrar_historico()

    # SAIR
    elif opcao == "6":

        print("\nSistema encerrado.")
        break

    # OPÇÃO INVÁLIDA
    else:

        print("\nOpção inválida! Tente novamente.")
```

**Desafio Master**

```python
from collections import deque

# Criando a fila de atendimento
fila = deque()


def mostrar_fila():
    print("\n--- FILA DE ATENDIMENTO ---")

    if not fila:
        print("A fila está vazia.")
    else:
        for i, paciente in enumerate(fila, 1):
            print(f"{i}º - {paciente}")

    print("---------------------------")


while True:
    print("\n===== SISTEMA DE ATENDIMENTO MÉDICO =====")
    print("1 - Adicionar paciente normal")
    print("2 - Adicionar paciente prioritário")
    print("3 - Atender próximo paciente")
    print("4 - Mostrar fila")
    print("5 - Sair")

    opcao = input("Digite uma opção: ")

    # Paciente normal entra no final da fila
    if opcao == "1":
        nome = input("Digite o nome do paciente: ")

        fila.append(nome)

        print(f"\nPaciente {nome} entrou no final da fila.")

    # Paciente prioritário entra no início da fila
    elif opcao == "2":
        nome = input("Digite o nome do paciente prioritário: ")

        fila.appendleft(nome)

        print(f"\nPaciente prioritário {nome} foi colocado no início da fila.")

    # Remove e atende o primeiro paciente
    elif opcao == "3":
        if fila:
            paciente = fila.popleft()
            print(f"\nAtendendo o paciente: {paciente}")
        else:
            print("\nNão há pacientes na fila.")

    # Mostra todos os pacientes
    elif opcao == "4":
        mostrar_fila()

    # Encerra o programa
    elif opcao == "5":
        print("\nSistema encerrado.")
        break

    else:
        print("\nOpção inválida! Digite uma opção de 1 a 5.")
```
