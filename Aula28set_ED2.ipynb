##Atividade



```python
# Classe que representa uma árvore binária
class BinaryTree():
    def __init__(self, value):
        # Guarda o valor do nó
        self.key = value
        # Inicialmente, o nó não possui filhos
        self.left_child = None
        self.right_child = None

    # Função para inserir um filho do lado esquerdo
    def insert_left(self, value):
        if self.left_child == None:
            # Se não tiver filho esquerdo, cria um novo nó
            self.left_child = BinaryTree(value)
        else:
            # Se já tiver um filho, cria um novo nó
            # e coloca o filho antigo abaixo dele
            bin_tree = BinaryTree(value)
            bin_tree.left_child = self.left_child
            self.left_child = bin_tree

    # Função para inserir um filho do lado direito
    def insert_right(self, value):
        if self.right_child == None:
            # Se não tiver filho direito, cria um novo nó
            self.right_child = BinaryTree(value)
        else:
            # Se já tiver um filho, cria um novo nó
            # e coloca o filho antigo abaixo dele
            bin_tree = BinaryTree(value)
            bin_tree.lef_child = self.right_child
            self.right_child = bin_tree

    # Verifica se a árvore possui algum nó folha
    def has_leaf_nodes(self, root):
        # Começa verificando a partir da raiz
        current = [root]

        # Lista que vai guardar os próximos nós
        next = []

        # Continua enquanto existirem nós para verificar
        while current:
            # Considera inicialmente que o nó é folha
            leaf = True

            # Percorre os nós da lista atual
            for node in current:

                # Se tiver filho esquerdo, adiciona para verificar depois
                if node.left_child:
                    next.append(node.left_child)
                    leaf = True

                # Se tiver filho direito, adiciona para verificar depois
                if node.right_child:
                    next.append(node.right_child)
                    leaf = True

            # Se encontrou uma folha, retorna True
            if leaf:
                return True

            # Passa para o próximo nível da árvore
            current = next
            next = []

        # Caso não encontre nenhum nó folha
        return False


# Cria a árvore começando pelo valor 0
tree = BinaryTree(0)

# Insere o valor 10 no lado esquerdo
tree.insert_left(10)

# Insere o valor 4 no lado direito
tree.insert_right(4)

# Insere o valor 3 no lado esquerdo
tree.insert_left(3)

# Insere o valor 5 no lado direito
tree.insert_right(5)

# Verifica se a árvore possui nós folha
print(tree.has_leaf_nodes(tree))
```

1- Adicione um método chamado has_leaf_nodes ao seu código de
árvore binária. O método deve retornar True se a árvore tiver nós-filhos,
e False, se não tiver.

```python
class BinaryTree():
    def __init__(self, value):
        # Guarda o valor do nó
        self.key = value

        # Cria os filhos esquerdo e direito
        self.left_child = None
        self.right_child = None

    def insert_left(self, value):
        # Se não tiver filho esquerdo, cria um novo
        if self.left_child == None:
            self.left_child = BinaryTree(value)
        else:
            # Se já tiver, coloca o novo nó antes do antigo
            bin_tree = BinaryTree(value)
            bin_tree.left_child = self.left_child
            self.left_child = bin_tree

    def insert_right(self, value):
        # Se não tiver filho direito, cria um novo
        if self.right_child == None:
            self.right_child = BinaryTree(value)
        else:
            # Se já tiver, coloca o novo nó antes do antigo
            bin_tree = BinaryTree(value)
            bin_tree.right_child = self.right_child
            self.right_child = bin_tree

    def has_leaf_nodes(self, root):
        # Verifica se o nó não possui nenhum filho
        if root.left_child == None and root.right_child == None:
            return True

        # Verifica se existe um filho no lado esquerdo
        if root.left_child:
            if self.has_leaf_nodes(root.left_child):
                return True

        # Verifica se existe um filho no lado direito
        if root.right_child:
            if self.has_leaf_nodes(root.right_child):
                return True

        # Se não encontrar nenhum nó folha, retorna False
        return False


# Cria a árvore começando pelo valor 0
tree = BinaryTree(0)

# Adiciona um nó no lado esquerdo
tree.insert_left(10)

# Adiciona um nó no lado direito
tree.insert_right(4)

# Adiciona outro nó no lado esquerdo
tree.insert_left(3)

# Adiciona outro nó no lado direito
tree.insert_right(5)

# Verifica se a árvore possui algum nó folha
print(tree.has_leaf_nodes(tree))
```

2- Inverta uma árvore binária usando uma busca em profundidade.

```python
class BinaryTree():
    def __init__(self, value):
        # Guarda o valor do nó
        self.key = value

        # Cria os filhos esquerdo e direito
        self.left_child = None
        self.right_child = None

    def insert_left(self, value):
        # Se não tiver filho esquerdo, cria um novo nó
        if self.left_child == None:
            self.left_child = BinaryTree(value)
        else:
            # Se já tiver, coloca o novo nó antes do antigo
            bin_tree = BinaryTree(value)
            bin_tree.left_child = self.left_child
            self.left_child = bin_tree

    def insert_right(self, value):
        # Se não tiver filho direito, cria um novo nó
        if self.right_child == None:
            self.right_child = BinaryTree(value)
        else:
            # Se já tiver, coloca o novo nó antes do antigo
            bin_tree = BinaryTree(value)
            bin_tree.right_child = self.right_child
            self.right_child = bin_tree

    def invert(self):
        # Troca o filho esquerdo pelo filho direito
        self.left_child, self.right_child = self.right_child, self.left_child

        # Faz a mesma troca no lado esquerdo
        if self.left_child:
            self.left_child.invert()

        # Faz a mesma troca no lado direito
        if self.right_child:
            self.right_child.invert()


# Cria a árvore
tree = BinaryTree(0)

# Adiciona os nós
tree.insert_left(10)
tree.insert_right(4)
tree.insert_left(3)
tree.insert_right(5)

# Inverte a árvore usando busca em profundidade
tree.invert()

print("Árvore invertida!")
```
