**Exercício 1 - Referências**

```python
a = [10, 20, 30]
b = a

b.append(40)

print("a:", a)
print("b:", b)
print("id(a):", id(a))
print("id(b):", id(b))
```

 **Qual será a saída?**

a: [10, 20, 30, 40]

b: [10, 20, 30, 40]

id(a): <algum número, ex: 140234567890>

id(b): <mesmo número de a>

**A e B representam o mesmo objeto?**

Sim. Ambos apontam para o mesmo objeto na memória.

**Por quê?**

Em Python, b = a não copia a lista — apenas faz b apontar para o mesmo objeto que a na memória. Como listas são mutáveis, a e b são duas referências para o mesmo objeto. Por isso, alterar b (com append, por exemplo) também altera a.

**O que aconteceria com b = a.copy()?**

a.copy() cria uma lista nova e separada, com seu próprio espaço na memória. Agora a e b são objetos diferentes (id(a) != id(b)), então mexer em um não afeta o outro.




**Exercício 2 — Construindo uma cadeia**

```python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.next = None


# Criando os nós
n1 = Node("A")
n2 = Node("B")
n3 = Node("C")

# Conectando: A → B → C → None
n1.next = n2
n2.next = n3
n3.next = None  # já é None por padrão, mas fica explícito

# Percorrendo a estrutura
atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.next
```

**Exercício 3 — Depuração**

```python
n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
```

**Qual é o problema?**

Dentro do while, o código nunca atualiza a variável atual, ela continua sempre apontando para n1.

**Por que o programa não termina?**

Porque a condição do laço é atual is not None, e como atual nunca muda, ela permanece n1 para sempre. Isso gera um loop infinito, imprimindo 10 (o valor de n1) repetidamente, sem nunca avançar para n2 ou n3.

**Qual linha deve ser acrescentada?**

Dentro do while, é preciso avançar para o próximo nó:  
atual = atual.proximo

**Qual será a saída depois da correção?**

10

20

30


**Exercício 4 - Problema real — Sistema de atendimento de uma clínica**

```python
from dataclasses import dataclass
from collections import deque


@dataclass
class Paciente:
    nome: str
    idade: int
    prioridade: str  # "Normal" ou "Prioridade"


class FilaAtendimento:
    def __init__(self):
        self.fila_prioridade = deque()
        self.fila_normal = deque()

    def adicionar_paciente(self):
        nome = input("Nome do paciente: ")
        idade = int(input("Idade: "))
        prioridade = input("Prioridade (Normal/Prioridade): ")

        paciente = Paciente(nome, idade, prioridade)

        if paciente.prioridade.lower() == "prioridade":
            self.fila_prioridade.append(paciente)
        else:
            self.fila_normal.append(paciente)

        print(f"Paciente {paciente.nome} adicionado com sucesso!\n")

    def listar_pacientes(self):
        todos = list(self.fila_prioridade) + list(self.fila_normal)
        if not todos:
            print("Nenhum paciente na fila.\n")
            return
        print("\nFila de atendimento:")
        for i, p in enumerate(todos, start=1):
            print(f"{i}. {p.nome} — {p.idade} anos — {p.prioridade}")
        print()

    def atender_primeiro(self):
        if self.fila_prioridade:
            paciente = self.fila_prioridade.popleft()
        elif self.fila_normal:
            paciente = self.fila_normal.popleft()
        else:
            print("Não há pacientes para atender.\n")
            return

        print(f"Atendendo: {paciente.nome} — {paciente.idade} anos — {paciente.prioridade}\n")

    def esta_vazia(self):
        if len(self.fila_prioridade) == 0 and len(self.fila_normal) == 0:
            print("A fila está vazia.\n")
        else:
            print("A fila NÃO está vazia.\n")

    def quantidade_pacientes(self):
        total = len(self.fila_prioridade) + len(self.fila_normal)
        print(f"Quantidade de pacientes na fila: {total}\n")


def menu():
    clinica = FilaAtendimento()

    while True:
        print("===== SISTEMA DE ATENDIMENTO - CLÍNICA =====")
        print("1 - Adicionar paciente")
        print("2 - Listar pacientes")
        print("3 - Atender primeiro paciente")
        print("4 - Verificar se a fila está vazia")
        print("5 - Quantidade de pacientes")
        print("0 - Sair")

        opcao = input("Escolha: ")

        if opcao == "1":
            clinica.adicionar_paciente()
        elif opcao == "2":
            clinica.listar_pacientes()
        elif opcao == "3":
            clinica.atender_primeiro()
        elif opcao == "4":
            clinica.esta_vazia()
        elif opcao == "5":
            clinica.quantidade_pacientes()
        elif opcao == "0":
            print("Encerrando o sistema...")
            break
        else:
            print("Opção inválida. Tente novamente.\n")


if __name__ == "__main__":
    menu()
```
